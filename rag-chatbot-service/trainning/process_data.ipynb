```python
import pandas as pd
from tqdm import tqdm
import os
import numpy as np
from config.setting import HCM_KEYWORD
from features.feature_engineering import build_features
from utils.helper import get_data_path, get_project_root

import requests
```

```python
input_path = get_data_path("ho_chi_minh_pricing.csv")
# "nhatot.csv"
# "ho_chi_minh_pricing" "vietnam_housing_dataset"
df = pd.read_csv(input_path)
```

```python
df["property_type"] = np.where(
    df["frontage"] > 0,
    "Nhà mặt phố, mặt tiền",
    np.where(
        df["access_road"] > 0,
        "Nhà ngõ, hẻm",
        "Unknown"
    )
)
df["property_feature"] = np.where(
    df["access_road"].isna(),
    "Unknown",
    np.where(
        df["access_road"] > 5,
        "Hẻm xe hơi",
        "Hẻm nhỏ"
    )
)
```

```python
output_path = get_data_path("ho_chi_minh_pricing_new.csv")
df = df.drop(["frontage", "access_road"], axis=1)
if os.path.exists(output_path):
    df.to_csv(output_path, mode="a", header=False, index=False, encoding="utf-8-sig")
else:
    df.to_csv(output_path, index=False, encoding="utf-8-sig")
```
